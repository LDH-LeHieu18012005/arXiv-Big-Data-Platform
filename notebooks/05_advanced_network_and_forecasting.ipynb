# 05 — Thống kê tác giả, hồi quy xu hướng từ khóa và thời gian thực thi

Dự báo tối đa 3 năm chỉ được công bố khi R² ≥ 0,5; MAE trên 3 năm holdout được báo cáo kèm để kiểm tra chéo. Kết quả chỉ áp dụng cho các shard đã tải, không đại diện toàn arXiv. Không khẳng định dự báo tới 2030.
